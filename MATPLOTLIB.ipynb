```python
import matplotlib.pyplot as plt
x=[10,200,30,400]
y=[50,60,70,80]
plt.plot(x,y)  #plot is used to draw
plt.show() #Show is used to display the output
```

```python
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr']
sales=[20,35,47,89]
plt.plot(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr']
sales=[20,35,47,89]
profit=[1,5,8,9]
plt.plot(months,sales,label='Sales') #label is to give names for each plot
plt.plot(months,profit,label='Profit')
plt.legend() #legend is to display label
plt.show()
```

```python
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr']
sales=[20,35,47,89]
plt.plot(months,sales)
plt.xlabel('months') #give name to x axis
plt.ylabel('sales')  #give name to y axis
plt.title('Sales Report') #give name to graph
plt.grid(True) #to give grid lines
plt.plot(months,sales,marker='o',color='red',linestyle=':')
plt.show()
```

```python
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr']
sales=[20,35,47,89]
plt.bar(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
study_hours=[1,2,3,4,5,6]
marks=[45,67,89,90,53,60]
plt.scatter(study_hours,marks)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
import matplotlib.pyplot as plt
marks=[45,67,89,90,53,60]
plt.hist(marks)
plt.title('Marks Distribution')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr']
sales=[20,35,47,89]
plt.pie(sales,labels=months)
plt.title('Sales by Month')
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,1) #fig represents whole graph
ax[0].plot([1,2,3,4],[10,20,30,40]) #ax represents individual graphs
ax[1].bar(['A','B','C','D'],[40,30,20,10])
plt.show()
```
